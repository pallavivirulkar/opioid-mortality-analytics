# Exploratory Analysis — Opioid Mortality Hotspots

Quick-look notebook for exploring the mortality and prescribing data outside the dashboard. Uses the same `src/` modules as `app/dashboard.py`, so anything validated here applies directly to the app.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import plotly.express as px

from src.data_processing import load_mortality_data, load_prescribing_data, data_source_status
from src.time_series import national_trend, decompose_series, compute_hotspot_scores
from src.cross_reference import merge_mortality_and_prescribing, overall_correlation, correlation_by_year

data_source_status()
```

## Load data

Falls back to `sample_data/` automatically if no real CDC/CMS exports are in `data/raw/`.

```
mortality = load_mortality_data()
prescribing = load_prescribing_data()
mortality.head()
```

## National trend

```
trend = national_trend(mortality)
px.line(trend, x="year", y="national_rate", markers=True, title="National opioid mortality rate")
```

## Hotspot leaderboard

```
hotspots = compute_hotspot_scores(mortality)
hotspots.head(10)
```

## Cross-reference with prescribing data

```
merged = merge_mortality_and_prescribing(mortality, prescribing)
print(overall_correlation(merged))
correlation_by_year(merged)
```

```
px.scatter(merged, x="opioid_prescribing_rate", y="crude_rate", color="state", facet_col="year", facet_col_wrap=4, trendline="ols", height=700, title="Mortality vs prescribing rate by year")
```
